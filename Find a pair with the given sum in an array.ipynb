```python
list = []
arr = int(input("enter no of elements for given array"))
for i in range (0, arr):
    element = int(input("enter values to be added"))
    list.append(element)
print(list)
val = int(input("enter pair sum u want to find"))
length = len(list) 
for i in range (length):
    for j in range (i+1, length):
        if (list[i]+list[j]) == val:
            print("pair found", (list[i],list[j]))
else:
    print("not found")
```
