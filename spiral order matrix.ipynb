```python
r = int(input("enter no of rows"))
c = int(input("enter no of columns"))
matrix = []
for i in range (r):
    a = []
    for j in range (c):
          a.append(int(input("enter the value to be added")))
    matrix.append(a)
for i in range (r):
    for j in range (c):
        print(matrix[i][j], end = " ")
    print()
```
