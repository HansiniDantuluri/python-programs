```python
list = []
arr = int(input("enter no of elements for given array"))
for i in range (0, arr):
    element = int(input("enter values to be added"))
    list.append(element)
print(list)
length = len(list) 
for i in range (length):
    for j in range (i+1, length):
        if (list[i] == list[j]):
            print("the duplicate element is : ", (list[i]))
else:
    print("not found")
```
